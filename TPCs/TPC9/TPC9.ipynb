```python
from gensim.models import Word2Vec
import re
import spacy

nlp = spacy.load("pt_core_news_sm")
documentos=["Harry_Potter_Camara_Secreta-br.txt","Harry_Potter_e_A_Pedra_Filosofal.txt"]
sentences = []
for doc in documentos:
    f = open(doc,"r", encoding="utf8") 
    texto = f.read()
    texto = re.sub(r'\f', '', texto)
    doc = nlp(texto)
    for sent in doc.sents:
        # Get the text of each token, lowercased, filtering out punctuation
        tokens = [token.text.lower() for token in sent if not token.is_punct and not token.is_space]
        if len(tokens) > 0:
            sentences.append(tokens)

model = Word2Vec(sentences, vector_size=100, window=5, min_count=2, sg=0, epochs=5, workers=3)
```

```python
result=model.wv.most_similar(positive=['harry','sonserina'],negative=['draco'])
print(result)
```

```python
model.wv.most_similar("harry")
```

```python
print(model.wv.similarity("harry","rony"))
print(model.wv.similarity("harry","hermione"))
print(model.wv.similarity("harry","sonserina"))
print(model.wv.similarity("harry","grifinória"))
```

```python
result=model.wv.most_similar(positive=['harry','sonserina'],negative=['draco'])
print(result)
```
